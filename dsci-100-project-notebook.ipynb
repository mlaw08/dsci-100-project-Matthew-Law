```python
import altair as alt
alt.Chart(WMM_tidy).mark_bar().encode(
    x=alt.X("reg_wins_pct").bin(maxbins=100).title("Regular Season Win Percentage (%)"),
    y=alt.Y("count()").title("Number of Teams")
).properties(
    title="Regular Season Win Percentage")
```

```python
alt.Chart(WMM_tidy).mark_bar().encode(
    x=alt.X("tourney_finish").title("Tournament Finish"),
    y=alt.Y("count()").title("Number of Teams")
).properties(
    title="Distribution of Tournament Finishes")
```
